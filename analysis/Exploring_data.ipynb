```python
df["price"] = df["price"].str.replace(",", "")
```

```python
df["rating_nums"] = (
    df["rating_nums"]
    .str.replace(" Ratings", "", regex=False)
    .str.replace("No rating", "0", regex=False)
    .astype("Int64")
)
```
